```python
X=df[['GrLivArea']]
Y=df['SalePrice']
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=30)
```

#### SK Learn Simple Linear regression

```python
Lr=LinearRegression()
```

```python

Lr.fit(x_train,y_train)
```

```python
print(f'The value of intercept is:{Lr.intercept_}')
print(f'The value of slope is:{Lr.coef_}')
```

#### Sk learn Ridge regression

```python
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
sacler=StandardScaler()
x_train_scaled=sacler.fit_transform(x_train)
x_test_scaled=sacler.transform(x_test)
ridge=Ridge(alpha=0.1)
ridge.fit(x_train_scaled,y_train)
print(f'The value of intercept is:{ridge.intercept_}')
print(f'The value of slope is:{ridge.coef_}')
```

```python
ry_pred=ridge.predict(x_test_scaled)
r2=r2_score(y_test,ry_pred)
print(f'The R2 Score of Simple regression is :{r2}')
```

#### Ridge Regression Code From Scratch

```python
class MyRidge:
    def __init__(self):
        self.alpha=0.1
        self.coef_=0
        self.intercept_=1
        
    def fit(self,x,y,alpha=0.1):
        x=x.ravel()
        self.alpha=alpha
        self.den=0
        self.num=0
        for i in range(x.shape[0]):
            self.den=self.den+(y.iloc[i]-y.mean())*(x[i]-x.mean())
            self.num=self.num+(x[i]-x.mean())*(x[i]-x.mean())
        
        self.coef_=self.den/(self.num+self.alpha)
        self.intercept_=y.mean()-(self.coef_*x.mean())
    
    def predict(self,x_test):
        x_test=x_test.ravel()
        return (x_test*self.coef_)+self.intercept_

# outside the class now:
ridge=MyRidge()
ridge.fit(x_train_scaled,y_train,alpha=50)
y_pred=ridge.predict(x_test_scaled)
r2=r2_score(y_test,y_pred)
print(f'The R2 Score of Simple regression is :{r2}')
```
